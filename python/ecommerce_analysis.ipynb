# Olist E-commerce Revenue & Customer Analytics

This notebook is the presentation-oriented companion to `build_analysis_tables.py`. Run the pipeline first to create the processed CSV files.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..')
PROCESSED = ROOT / 'data' / 'processed'
pd.set_option('display.max_columns', None)
```

## 1. Executive KPI Summary

```python
kpis = pd.read_csv(PROCESSED / 'kpi_summary.csv')
kpis
```

## 2. Monthly Revenue Trend

```python
monthly = pd.read_csv(PROCESSED / 'monthly_kpis.csv', parse_dates=['order_month'])
monthly.plot(x='order_month', y='merchandise_revenue', figsize=(10,5), title='Monthly Merchandise Revenue')
plt.ylabel('Revenue (BRL)')
plt.tight_layout()
plt.show()
```

## 3. Top Product Categories

```python
category = pd.read_csv(PROCESSED / 'category_performance.csv')
category.head(10).sort_values('merchandise_revenue').plot.barh(x='category', y='merchandise_revenue', figsize=(9,6), title='Top 10 Categories by Merchandise Revenue')
plt.tight_layout()
plt.show()
```

## 4. Customer Segments

```python
rfm = pd.read_csv(PROCESSED / 'rfm_segment_summary.csv')
rfm
```

## 5. Cohort Retention

```python
cohort = pd.read_csv(PROCESSED / 'cohort_retention.csv', parse_dates=['cohort_month'])
retention_matrix = cohort.pivot(index='cohort_month', columns='cohort_index', values='retention_rate_pct')
retention_matrix.head()
```

## 6. Delivery & Customer Satisfaction

```python
service = pd.read_csv(PROCESSED / 'delivery_review_analysis.csv')
service.groupby('is_late')['review_score'].agg(['count','mean','median'])
```
